```python
'''
WAPP to explain working with bokeh line graph using Annotation legends.
(plot different types of plots using bokeh)
'''

from bokeh.plotting import figure, show

x=[1,2,3,4,5]
y=[6,7,2,4,5]

p=figure(title="Interactive Line Graph\nADITI A MURTHY\n1JS24CS007", x_axis_label='X-axis', y_axis_label='Y-axis')

p.line(x,y, legend_label="Line", line_width=2, color='lime')

p.annular_wedge(x=5, y=5, inner_radius=0.2, outer_radius=0.4,start_angle=45, end_angle=135,line_color="thistle", fill_color="lightcyan")

show(p)
```
