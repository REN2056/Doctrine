```python
'''
Write a Python program for creating maps using Plotly Libraries.
'''

import plotly.express as px

# India Locations
india_locations = ["Delhi", "Bangalore", "Arunachal Pradesh"]
india_latitudes = [28.7041, 12.9629, 28.2180]
india_longitudes = [77.1025, 77.5775, 94.7278]

# USA Locations
usa_locations = ["New York", "Los Angeles", "Chicago", "San Francisco"]
usa_latitudes = [40.7128, 34.0522, 41.8781, 37.7749]
usa_longitudes = [-74.0060, -118.2437, -87.6298, -122.4194]

# Create World Map
fig = px.scatter_geo()

# USA Points
fig.add_scattergeo(
    lat=usa_latitudes,
    lon=usa_longitudes,
    text=usa_locations,
    marker=dict(size=8, color="red"),
    name="USA"
)

# India Points
fig.add_scattergeo(
    lat=india_latitudes,
    lon=india_longitudes,
    text=india_locations,
    marker=dict(size=8, color="blue"),
    name="India"
)

# Map Styling
fig.update_geos(
    visible=True,
    showland=True,
    landcolor="LightGreen",
    showocean=True,
    oceancolor="LightBlue",
    showlakes=True,
    lakecolor="LightBlue"
)

# Layout
fig.update_layout(
    title="World Map Showing Locations in India and USA",
    title_x=0.5
)

# Display Map
fig.show()
```
