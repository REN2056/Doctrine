```python
'''
WAPP to draw 3D plots using plotly library
'''

import plotly.graph_objects as go
import numpy as np

x=np.linspace(0,10,100)
y=np.linspace(0,10,100)
z=np.random.rand(100,100)

fig=go.Figure(data=[go.Scatter3d(x=x,y=y,z=z,mode='markers')])

fig.update_layout(title="3D Scatter Plot\nADITI A MURTHY\nJS24CS007",scene=dict(xaxis_title='X AXIS',yaxis_title='Y AXIS',zaxis_title='Z AXIS'))

fig.show()



'''
HTML output you're seeing is Plotly's graph data printed to the terminal.
Plotly graphs do not display as windows by default when executed in a normal Python script (unless you add fig.show() AND have a browser environment).
Right now, your script is printing the figure as raw HTML, which is normal behavior for Plotly in a terminal.

'''
```
